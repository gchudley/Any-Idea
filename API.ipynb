```python
import requests
import os
import pandas as pd

url = "https://data.cityofnewyork.us/resource/5zyy-y8am.json"
page_size = 50000
offset = 0
rows = []

while True:
    params = {
        "$where": "report_year=2024",
        "$order": ":id",
        "$limit": page_size,
        "$offset": offset,
    }
    r = requests.get(url, params=params)
    r.raise_for_status()
    batch = r.json()
    if not batch:
        break
    rows.extend(batch)
    offset += page_size

df = pd.DataFrame(rows)
print(len(df))

os.makedirs("data", exist_ok=True)
df.to_csv("data/nyc_2024_data.csv", index=False)
```
